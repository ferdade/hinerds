```python
lista_retirar = ['Prefiro não informar','Outra','Indígena']
```

```python
dados = dados[~dados['COR/RACA/ETNIA'].isin(lista_retirar)]
```

```python
dados['nao_branca'] = dados['COR/RACA/ETNIA'].apply(lambda x: 1 if x != 'Branca' else 0)
```

```python
dados['Quanto tempo de experiência na área de TI/Engenharia de Software você teve antes de começar a trabalhar na área de dados?'].value_counts()
```

```python
dados['Tempo de expêriencia'] = dados['Quanto tempo de experiência na área de TI/Engenharia de Software você teve antes de começar a trabalhar na área de dados?'].str.extract(r'(\d+)')
```

```python
dados['Tempo de expêriencia']
```

```python
dados['NUMERO DE FUNCIONARIOS'].value_counts()
```

```python
dados['NUMERO DE FUNCIONARIOS'] = dados['NUMERO DE FUNCIONARIOS'].str.replace('.','')
```

```python
dados['NUMERO DE FUNCIONARIOS'] = dados['NUMERO DE FUNCIONARIOS'].str.extract(r'(\d+)')
```

```python
dados['NUMERO DE FUNCIONARIOS'].value_counts()
```

```python
dados['Qual o principal motivo da sua insatisfação com a empresa atual?'].value_counts()
```

```python
dados.loc[dados['Qual o principal motivo da sua insatisfação com a empresa atual?'].notnull(), 'Qual o principal motivo da sua insatisfação com a empresa atual?'].apply(lambda x: 1 if 'Salário' in x else 0)
```

```python
mask = dados['Qual o principal motivo da sua insatisfação com a empresa atual?'].notnull()
dados.loc[mask, 'Insatisfação'] = dados.loc[mask, 'Qual o principal motivo da sua insatisfação com a empresa atual?'].str.contains('Salário').astype(int)
```

```python
dados['Insatisfação'].value_counts()
```
